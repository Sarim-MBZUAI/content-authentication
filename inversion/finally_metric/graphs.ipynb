# SD3.5 unconditioned one

```python
import numpy as np
import json
import matplotlib.pyplot as plt
from scipy import stats

def process_metrics(file_path):
    # Read the JSONL file
    metrics = []
    with open(file_path, 'r') as f:
        for line in f:
            data = json.loads(line)
            metrics.append(data['metrics']['combined_score'])
    
    return np.array(metrics)

def create_comparison_plot(file_path1, file_path2, label1="Original", label2="Comparison"):
    # Process both files
    metrics1 = process_metrics(file_path1)
    metrics2 = process_metrics(file_path2)

    # Create the plot
    plt.figure(figsize=(12, 6))

    # Create histograms
    plt.hist(metrics1, bins='auto', density=True, alpha=0.5, color='skyblue', label=f'{label1} Histogram')
    plt.hist(metrics2, bins='auto', density=True, alpha=0.5, color='lightgreen', label=f'{label2} Histogram')

    # Add kernel density estimates
    kde1 = stats.gaussian_kde(metrics1)
    kde2 = stats.gaussian_kde(metrics2)
    x_range = np.linspace(min(min(metrics1), min(metrics2)), max(max(metrics1), max(metrics2)), 100)
    plt.plot(x_range, kde1(x_range), 'b-', lw=2, label=f'{label1} Density')
    plt.plot(x_range, kde2(x_range), 'g-', lw=2, label=f'{label2} Density')

    # Add mean lines
    plt.axvline(np.mean(metrics1), color='blue', linestyle='dashed', linewidth=2)
    plt.axvline(np.mean(metrics2), color='green', linestyle='dashed', linewidth=2)

    # Customize plot
    plt.grid(True, linestyle='--', alpha=0.7)
    plt.title('Distribution of Combined Scores', fontsize=12, pad=20)
    plt.xlabel('Combined Score', fontsize=10)
    plt.ylabel('Density', fontsize=10)
    plt.legend()

    # Print statistics
    print(f"\n{label1} Statistics:")
    print(f"Mean: {np.mean(metrics1):.4f}")
    print(f"Median: {np.median(metrics1):.4f}")
    print(f"Standard Deviation: {np.std(metrics1):.4f}")
    print(f"Number of images: {len(metrics1)}")

    print(f"\n{label2} Statistics:")
    print(f"Mean: {np.mean(metrics2):.4f}")
    print(f"Median: {np.median(metrics2):.4f}")
    print(f"Standard Deviation: {np.std(metrics2):.4f}")
    print(f"Number of images: {len(metrics2)}")

    plt.tight_layout()
    plt.show()

# Usage example:
create_comparison_plot('/shared/shashmi/inversion/new_metric/finally_metric/SD3.5_uncondtioned_image_fake_inv_fake_metrics.jsonl', '/shared/shashmi/inversion/new_metric/finally_metric/SD3.5_uncondtioned_image_real_inv_real_metrics.jsonl', 
                      label1="fake_inv_fake", label2="real_inv_real")
```
